Let's look at our 3rd business question: do perfumes that are harder to find sell out more quickly?

```python
pivot_table = df.pivot_table(values='sold', index=pd.cut(df['available'], bins=5), aggfunc='mean')

plt.figure(figsize=(8, 5))
sns.heatmap(pivot_table, annot=True, cmap='coolwarm', fmt=".1f", linewidths=0.5)
plt.xlabel("Stock Available (Binned)")
plt.ylabel("Units Sold")
plt.title("Average Sales Across Stock Levels")
plt.show()
```

Result:
Perfumes with lower stock levels often sell out faster, indicating higher demand.
However, a consistent stock level is essential to maintain continuous sales.

Let's examine our 4th business question, which is: Which locations generate the most sales?

The location with the highest sales is shown in the question above. Let's move on to our best-selling fragrance and conclude our 5th business question.

```python
plt.figure(figsize=(10, 5))
type_sales = df.groupby('type')[['sold']].sum().sort_values(by='sold', ascending=False).head(5)
sns.barplot(x=type_sales.index, y=type_sales['sold'], palette='viridis')
plt.xlabel("Fragrance Type")
plt.ylabel("Units Sold")
plt.title("Top 5 Selling Fragrance Types")
plt.xticks(rotation=45, ha='right')
plt.show()
```

We may now examine our 6th business question, which asks which scents have the highest sales-to-stock ratio, suggesting high demand relative to availability, after observing the best-selling fragrance.

```python
df['sales_to_stock_ratio'] = df['sold'] / (df['available'] + 1)
top_selling_performance = df[['title', 'sales_to_stock_ratio']].sort_values(by='sales_to_stock_ratio', ascending=False).head(10)
plt.figure(figsize=(12, 6))
sns.barplot(x='sales_to_stock_ratio', y='title', data=top_selling_performance, palette='magma')
plt.xlabel("Sales-to-Stock Ratio")
plt.ylabel("Perfume Title")
plt.title("Top 10 Perfumes by Sales-to-Stock Ratio")
plt.show()
```

According to our results, Calvin Klein Cologne is the best-selling perfume. It also has the largest stock sales ratio, making it the one with the highest stock sales.The highest sales-to-stock ratio indicates perfumes that are in high demand but have limited availability

Let's move on to our 7th business question, which asks how rival price impacts sales.

```python
# Group by brand and calculate total sales to find the top 20 brands
brand_sales = df.groupby('brand')['sold'].sum().sort_values(ascending=False)
top_20_brands = brand_sales.head(20).index.tolist()

# Filter the dataset to only include the top 20 brands
top_20_brand_data = df[df['brand'].isin(top_20_brands)]

# Create a boxplot to visualize the pricing strategy of the top 20 brands
plt.figure(figsize=(12, 6))
sns.boxplot(x='brand', y='price', data=top_20_brand_data)
plt.xticks(rotation=90)
plt.xlabel("Brand")
plt.ylabel("Price ($)")
plt.title("Pricing Strategies of Top 20 Brands")
plt.show()
```
